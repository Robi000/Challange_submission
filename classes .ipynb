# 🐍 Challenge: Engineering a Custom `Range` Class

The goal is to re-create Python’s built-in `range` object from scratch. This isn't just about loops; it's about understanding **memory efficiency** and **lazy evaluation**.

---

### 🚀 The Core Philosophy
Unlike a list, a `range` does not store all its numbers in memory. It calculates them **on demand**. Your class must behave like a list without ever actually *becoming* one.

---

### 🛠️ The Technical Hurdles

#### 1. Lazy Evaluation & Memory Efficiency
* **The Problem:** Storing `range(1_000_000)` in a list would consume significant RAM.
* **The Goal:** Design the class to calculate each value only when requested.
* **Constraint:** You cannot use `list` or `tuple` to store the sequence.

#### 2. Dynamic Parameter Handling
Your class must be flexible enough to handle three different call signatures:
* `Range(stop)` → `start=0`, `step=1`
* `Range(start, stop)` → `step=1`
* `Range(start, stop, step)` → full control

#### 3. Instant Length Calculation (`__len__`)
You cannot iterate through the sequence to count them. You must use a mathematical formula to return the length in $O(1)$ time.
> **Formula Hint:** > $$length = \max(0, \frac{stop - start + step - 1}{step})$$
> *(Note: This logic must be adjusted to handle both positive and negative steps!)*

#### 4. Smart Indexing (`__getitem__`)
Support `r[k]` access just like a list.
* **Direct Calculation:** Compute the $k$-th element mathematically: $start + (k \times step)$.
* **Negative Indices:** Support `r[-1]` by converting it to a positive index internally.
* **Validation:** Raise an `IndexError` if the index is out of bounds.

#### 5. Efficient Iteration (`__iter__`)
The class must be compatible with `for x in r:` loops.
* **The Twist:** Use a generator or an iterator to yield values one by one so the memory footprint remains near zero.

---

### ⚠️ Edge Cases to Master
| Case | Scenario | Expected Behavior |
| :--- | :--- | :--- |
| **Zero Step** | `Range(0, 10, 0)` | Raise `ValueError` |
| **Empty Range** | `Range(10, 0, 1)` | Length should be `0` |
| **Negative Step** | `Range(10, 0, -2)` | Should count down: `10, 8, 6, 4, 2` |
| **Out of Bounds** | `r[100]` on a small range | Raise `IndexError` |
