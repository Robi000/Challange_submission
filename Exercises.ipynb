````python
# efficent one with O(N) time and O(1) space usage
def minmax(data):
    itr  = iter(data)
    try:
        low = high = next(itr)
    except StopIteration:
        raise ValueError("have impty sequence")
    

    for value in itr:
        if low > value:
            low = value 
        if high < value:
            high = value 
    
    return (low, high)
````

# 🧮 Challenge: Sum of Squares

### **The Objective**
Calculate the cumulative total of squared integers within a specific range.

---

### **Task Description**
Write a Python function that accepts a single positive integer $n$ and returns the sum of the squares of all positive integers **smaller than** $n$.

### **Mathematical Representation**
The function should compute:
$$\sum_{i=1}^{n-1} i^2$$

---

### **Specifications**
* **Input**: A positive integer `n`.
* **Output**: An integer representing the sum ($1^2 + 2^2 + \dots + (n-1)^2$).
* **Edge Case**: If $n=1$, the result should be `0` (as there are no positive integers smaller than 1).

---

### **Example**
**Input:** `n = 4`  
**Logic:** $1^2 + 2^2 + 3^2 = 1 + 4 + 9$  
**Output:** `14`

````python
def Sqrless (value):
    total = 0
    while value > 0:
        value -= 1 
        total += value * value
    return total
````

````python
def test_sum_squares(func):
    """
    Tests the sum of squares function across multiple ranges.
    """
    scenarios = [
        (1, 0, "Minimal case (n=1)"),
        (2, 1, "n=2 (1^2)"),
        (4, 14, "n=4 (1+4+9)"),
        (5, 30, "n=5 (1+4+9+16)"),
        (10, 285, "n=10"),
        (20, 2470, "Larger range")
    ]

    # print(f"✨ Testing 'Sum of Squares' for robi...\n")
    # print(f"{'Input (n)':<10} | {'Status':<8} | {'Details'}")
    # print("-" * 45)

    for n, expected, desc in scenarios:
        result = func(n)
        status = "✅ PASS" if result == expected else "❌ FAIL"
        # print(f"{n:<10} | {status:<8} | {desc}")

# Usage:
test_sum_squares(Sqrless)
````

````python
# two faster methodes include square pyramidal number formula and comprehension syntax

def sqrless1 ( Value):
    Value = Value - 1 
    return (Value * (Value + 1 ) * (2* Value + 1 ))//6
````

````python
%%timeit
test_sum_squares(sqrless1)
````

# 🧩 Challenge: Sum of Odd Squares

### **The Objective**
Calculate the cumulative total of squares, but only for the **odd** integers in the sequence.

---

### **Task Description**
Write a Python function that accepts a single positive integer $n$ and returns the sum of the squares of all **odd** positive integers **smaller than** $n$.

### **Mathematical Representation**
The function should compute:
$$\sum_{i \in \{1, 3, 5, \dots < n\}} i^2$$

---

### **Specifications**
* **Input**: A positive integer `n`.
* **Output**: An integer representing the sum of odd squares.
* **Logic**: If $n=5$, you sum $1^2 + 3^2$ (5 is not included because it must be *smaller than* $n$).

---

### **Example**
**Input:** `n = 6`  
**Odd integers < 6:** `1, 3, 5`  
**Calculation:** $1^2 + 3^2 + 5^2 = 1 + 9 + 25$  
**Output:** `35`


````python
def sqrodd ( value ): 
    # print ([n for n in range (1,value) if  n % 2])
    return sum ([n ** 2 for n in range (1,value) if  n % 2])
````

````python
sqrodd (10)
````

````python
def test_sum_odd_squares(func):
    """
    Tests the sum of odd squares function with boundary and edge cases.
    """
    scenarios = [
        (1, 0, "n=1 (No positive odds < 1)"),
        (2, 1, "n=2 (Only 1^2)"),
        (3, 1, "n=3 (Only 1^2, 3 is not smaller than 3)"),
        (4, 10, "n=4 (1^2 + 3^2)"),
        (5, 10, "n=5 (1^2 + 3^2, 5 is not smaller than 5)"),
        (6, 35, "n=6 (1^2 + 3^2 + 5^2)"),
        (10, 165, "n=10 (1^2 + 3^2 + 5^2 + 7^2 + 9^2)"),
        (1, 0, "Edge case: n is 1"),
        (0, 0, "Edge case: n is 0"),
    ]

    print(f"🧪 Testing 'Sum of Odd Squares' for robi...\n")
    print(f"{'Input (n)':<10} | {'Status':<8} | {'Details'}")
    print("-" * 50)

    for n, expected, desc in scenarios:
        try:
            result = func(n)
            status = "✅ PASS" if result == expected else "❌ FAIL"
            print(f"{n:<10} | {status:<8} | {desc} (Got: {result})")
        except Exception as e:
            print(f"{n:<10} | 💥 ERROR  | {desc} ({e})")

# Usage: 
test_sum_odd_squares(sqrodd)
````

# 🎂 Challenge: The Birthday Paradox

### **The Objective**
Empirically verify the statistical claim that in a group of **23 people**, the probability of a shared birthday exceeds **50%**.

---

### **Task Description**
Design a Python program to simulate the "Birthday Paradox" through Monte Carlo experiments. The simulation should demonstrate that as the group size $n$ increases, the probability of at least one pair sharing a birthday rises significantly faster than intuition suggests.

### **The Mathematical Logic**
The probability $P(A)$ that at least two people share a birthday is calculated as:
$$P(A) = 1 - \frac{365!}{365^n (365-n)!}$$

---

### **Program Requirements**
1. **Experiment**: Create a function that generates $n$ random integers between 1 and 365.
2. **Detection**: Check if the generated list contains any duplicates.
3. **Simulation**: Run the experiment $10,000$ times for a specific $n$ to find the average probability.
4. **Validation**: Test for $n = 23$ and confirm if the result is $> 50\%$.

---

### 💡 **Example Output**
```text
Running 10,000 simulations for n=23...
Shared birthdays found in 5,073 trials.
Estimated Probability: 50.73%
Claim Verified: True

````python
def calculate_birthday_paradox ( num_people ):
    import random
    birthdays = [random.randint(1, 365) for _ in range(num_people)]
    return len(birthdays) != len(set(birthdays))
````

````python
average_trials = 100000
result = 0

for trial_No in range(1 , average_trials+1):
    if calculate_birthday_paradox(23):
        result  = (result * (trial_No - 1) + 1) / trial_No

result = result* 100

print(f"Estimated probability of shared birthdays in a group of 23 people: {result:.4f}%")
````
